```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#define funtion for EAR (effective annual rate)for a bond coupon

#remember define a function we give it ingerdients
#ingredients - (r=coupon p.a),(interval= the interval between periods),(periods_per_year = month,week,day etc)
def EAR(r,interval,periods_per_year):
  n=periods_per_year/interval
#this gives us the number of weeks between payments rather than
#directly the number of compounding periods
#followed by recipe(formula)
  ear=(((1+r/n)**n)-1) # ** is for exponents
  return ear
```

```python
#create a loop function to loop through periods for the EAR

#first create an empty list to add results to
ear_results=[]
#create range to loop through - weeks 1-52
for i in range(1,53):
  #using EAR function above 0.1(10%) and looping through i(weeks 1-52)
  # append to the list ear_results
  ear_results.append(EAR(0.1,i,52)) #this calls our function and stores the results
print(ear_results)
```

```python
# create a chart displaying the change in EAR as compunding periods decrease in frequency

#already have list for EAR results so;
#create a list for periods weeks 1-52
periods=list(range(1,53)) #range is a series of numbers from 1-52

#convert ear_results to a percentage for clarity
ear_results_pct=[x*100 for x in ear_results]
#now plot using plt.plot function for line/scatter
plt.plot(periods,ear_results_pct,linewidth=2,label='EAR')
plt.title('EAR of 10% p.a as period of payments increases from 1 week to 52 weeks')
plt.xlabel('Period (weeks)')
plt.ylabel('EAR(%)')
plt.grid(True,alpha=0.3)
plt.savefig('EAR_graph.png',dpi=150,bbox_inches='tight')
plt.show()
```
